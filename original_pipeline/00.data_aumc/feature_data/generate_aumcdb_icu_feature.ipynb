```python
import pandas as pd
import numpy as np
```

```python
# 临床特征标准的start是 time//60； end为 time//60 + 1 即入ICU的第几小时
df_fea = pd.read_csv("/public/hanl/jupyter_dir/database/AMUCdb/AUMCdb_suit_for_MIMIC/AUMCdb_finaldata_suit_for_mimic.csv") # 时间单位为小时
df_fea = df_fea.drop(["end", "if_del"],axis=1)
df_fea = df_fea.rename(columns={'admissionid':'stay_id',"start":"charttime"}) 
print(len(df_fea.stay_id.unique()))
df_fea = df_fea.rename(columns={'temperture':"temperature",'PaO2':'po2',"PaCO2":"pco2","FiO2":"fio2", "urine_output":"urineoutput", "gcs_score":"gcs", 
                                        "eyes_score":"gcs_eyes", 'motor_score':"gcs_motor", 'verbal_score':"gcs_verbal"})

# 去除异常值
df = df_fea
df.pt=df.pt*10
df.calcium=df.calcium*10
df.temperature[(df.temperature>=50)|(df.temperature<=30)]=np.nan #df.temperature[(df.temperature<50)&(df.temperature>30)].mean()
df.heart_rate[df.heart_rate>=200] = np.nan #df.heart_rate[df.heart_rate<200].mean()
df.resp_rate[df.resp_rate>=50]= np.nan #df.resp_rate[df.resp_rate<50].mean()
df.sbp[df.sbp>=160]= np.nan #df.sbp[df.sbp<160].mean()
df.mbp[(df.mbp>=110)|(df.mbp<=40)]= np.nan #df.mbp[(df.mbp<110)&(df.mbp>40)].mean()
df.bicarbonate[df.bicarbonate>=35]= np.nan #df.bicarbonate[df.bicarbonate<35].mean()
df.baseexcess[(df.baseexcess>=10)|(df.baseexcess<=-20)]=np.nan #df.baseexcess[(df.baseexcess<10)&(df.baseexcess>-20)].mean()
df.lactate[df.lactate>=25]=np.nan #df.lactate[df.lactate<25].mean()
df.pco2[df.pco2>=70]=np.nan #df.pco2[df.pco2<70].mean()
df.loc[df["spo2"]<10, "spo2"] *= 100
df.spo2[(df.spo2>=100)|(df.spo2<=80)]=np.nan #df.spo2[(df.spo2<100)&(df.spo2>80)].mean()
df.glucose[(df.glucose>=1000)|(df.glucose<=10)]=np.nan

df.po2[df.po2>=300]=np.nan #df.po2[df.po2 < 300].mean()
df.wbc[df.wbc>=50]=np.nan #df.wbc[df.wbc < 50].mean()
df.ptt[df.ptt>=120]= np.nan #df.ptt[df.ptt < 120].mean()
df.ph[(df.ph>=10)|(df.ph<=6)]=np.nan # df.ph[(df.ph<10)&(df.ph>6)].mean()
df.urineoutput[(df.urineoutput>=1000)|(df.urineoutput<=0)]=np.nan #df.urineoutput[(df.urineoutput<1000)&(df.urineoutput>0)].mean()
df.chloride[(df.chloride>=130)|(df.chloride<=80)]=np.nan #df.chloride[(df.chloride<130)&(df.chloride>80)].mean()
df.potassium[(df.potassium>=7)|(df.potassium<=3)]= np.nan #df.potassium[(df.potassium<7)&(df.potassium>3)].mean()
df_fea = df

df_fea.to_csv("aumc_icu_feature.csv",index=False)
```
